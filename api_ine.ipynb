```python
import requests
import pandas as pd
```

```python
#DESCARGA DATOS POBLACION RESIDENTE NACIDA EXTRANJERO. TOTAL
# API INE
url = "https://servicios.ine.es/wstempus/js/es/DATOS_TABLA/56943?tip=AM"

# Descargar JSON
response = requests.get(url)
data = response.json()

rows = []

for item in data:

    nombre = item.get("Nombre", "")

    # FILTROS
    if (
        "País Vasco" in nombre and
        "Extranjero" in nombre and
        "Total" in nombre
    ):

        if "Data" in item:

            for d in item["Data"]:

                rows.append({
                    "Serie": nombre,
                    "Fecha": d.get("Fecha"),
                    "Valor": d.get("Valor")
                })

# Crear DataFrame
df = pd.DataFrame(rows)

# Separar dimensiones en columnas
columnas = df["Serie"].str.split(".", expand=True)

# Ajustar nombres según estructura INE
df["Comunidad"] = columnas[0]
df["Edad"] = columnas[1]
df["Lugar_nacimiento"] = columnas[2]
df["Sexo"] = columnas[3]

# Reordenar columnas
df = df[
    [
        "Comunidad",
        "Edad",
        "Lugar_nacimiento",
        "Sexo",
        "Fecha",
        "Valor"
    ]
]

# Exportar Excel
archivo = "INE_Pais_Vasco_Extranjero.xlsx"

df.to_excel(archivo, index=False)

print(f"Archivo generado: {archivo}")
```
